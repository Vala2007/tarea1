### **Ejercicio 1**

![1](captura1.png)

```python
import numpy as np

def factorial(n): # Calcula el factorial de un número n
    resultado = 1
    for i in range(1, n + 1):
        resultado *= i
    return resultado

n = int(input("Ingrese el valor de n:")) 

fact = factorial(n) 
print(f"Factorial de {n}: {fact}")

V1 = np.random.randint(1, 11, size=n) # Genera un vector V1 de tamaño n con números aleatorios entre 1 y 10
V2 = np.random.randint(1, 11, size=n) # Genera un vector V2 de tamaño n con números aleatorios entre 1 y 10
V3 = np.where(V1 < V2, 0, 1) # Genera un vector donde cada elemento es 0 si el elemento correspondiente en V1
# es menor que el de V2, y 1 en caso contrario

matriz = np.concatenate((V1.reshape(-1, 1), V2.reshape(-1, 1), V3.reshape(-1, 1)), axis=1) 
# Crea una matriz de tamaño n x 3, primera columna es V1, segunda columna es V2, y tercera columna es V3
print(matriz, "\n")
```

### **Ejercicio 2**

![2](captura2.png)

```python
import numpy as np

m = int(input("Ingrese el número de filas (m): "))
n = int(input("Ingrese el número de columnas (n): "))

A = np.zeros((m, n), dtype=int) # Crea una matriz A de tamaño m x n inicializada con ceros

col = 0
while col < n:
# Columna aleatoria de tamaño m con valores entre -1 y 3, y la asigna a la columna correspondiente de la matriz A
    columna_aleatoria = np.random.randint(-1, 4, size=m)
    A[:, col] = columna_aleatoria
    col += 1
print("Matriz A original:", 2*"\n", A, "\n")

A_while = A.copy()
A_for = A.copy()

# Transformación de la matriz A, reemplazando los valores negativos por -1 y los no negativos por 5, usando while
i = 0
while i < m:
    j = 0
    while j < n:
        if A_while[i, j] < 0:
            A_while[i, j] = -1
        else:
            A_while[i, j] = 5
        j += 1
    i += 1
print("Matriz A transformada usando while:\n")
print(A_while, "\n")

# Transformación de la matriz A, reemplazando los valores negativos por -1 y los no negativos por 5, usando for
for i in range(m):
    for j in range(n):
        if A_for[i, j] < 0:
            A_for[i, j] = -1
        else:
            A_for[i, j] = 5
print("Matriz A transformada usando for:\n")
print(A_for, "\n")

print("Comparación de resultados:", np.array_equal(A_while, A_for), "\n")
```
